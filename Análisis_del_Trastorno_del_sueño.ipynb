##Planteamiento del problema

###¿Qué es un Transtorno del sueño?

Los trastornos del sueño son afecciones que afectan la calidad, duración o el patrón del sueño. Pueden incluir problemas para conciliar el sueño, mantenerse dormido, o experimentar conductas anormales durante el sueño.

Entre ellos:


* Insomnio.
* Transtornos respiratorios.
* Somnolencia.
* Sueño insuficiente inducido por el comportamiento.
* Ronquido.
* Narcolepsia.
* Parasomnias.
* Terrores nocturnos.
* Sonambulismo.

Vamos a explicar de manera breve en qué consiste cada una:

---

###Insomnio.
Es un trastorno del sueño muy común. Quien lo padece, tiene problemas para lograr conciliar el sueño, quedarse dormido o los dos.

Síntomas:

* Permanecer despierto por un largo tiempo antes de dormirse.
* Dormir solo por períodos cortos.
* Estar despierto durante gran parte de la noche.
* Sentirse como si no hubiera dormido nada.
* Despertarse demasiado temprano.

Posible tratamiento:

Los tratamientos pueden incluir cambios en el estilo de vida, terapia o medicamentos, según el tipo de insomnio (Agudo: de corta duración o crónico: de larga duración), el paciente y la gravedad del problema, siempre diagnosticado por:

* Los cambios en el estilo de vida, incluidos los buenos hábitos para dormir, a menudo ayudan a aliviar el insomnio agudo (a corto plazo).
* Un tipo de terapia, la terapia cognitiva-conductual, puede ayudar a aliviar la ansiedad relacionada con el insomnio crónico.
* Varios medicamentos también pueden ayudar a aliviar el insomnio y le permiten restablecer un horario de sueño regular.

---

###Transtornos Respiratorios.

En este grupo se incluyen aquellos trastornos del sueño que se caracterizan por una alteración de la respiración durante el sueño.

estos se dividen en varios tipos:

* Los síndromes de apnea central del sueño incluyen aquellos trastornos en los que el movimiento respiratorio está disminuido o ausente de una manera intermitente o cíclica, debido a una disfunción cardíaca o del sistema nervioso central.

* En el síndrome de apneas obstructivas del sueño se produce una obstrucción al flujo del aire en la vía aérea, de manera que en presencia de movimientos respiratorios, a menudo enérgicos para intentar superar la obstrucción, hay una inadecuada ventilación.

* Síndrome de apnea/hipopnea obstructiva del sueño (SAHOS) Consiste en la presencia de apnea, ronquido, durante el sueño, de episodios repetidos de obstrucción, completa en el caso de las apneas y parcial en el caso de las hipopneas.

Sintomas:

Un signo frecuentemente asociado son los ronquidos muy intensos, que suelen producirse entre las apneas.

---

###Somnolencia.

La somnolencia diurna excesiva, es un trastorno del sueño que se refiere a la necesidad de dormir involuntariamente frecuente e incontrolable que se presenta durante el día; a pesar de haber dormido tiempo suficiente.

Éste es el síntoma principal de la narcolepsia. Contrariamente a los insomnios, los pacientes se quejan rara vez de la somnolencia diurna.

Esta sensación de sueño es habitual en cerca de un 8% de la población.

---

###Sueño insuficiente inducido por el comportamiento

Se trata de un trastorno del sueño que se caracteriza por somnolencia excesiva diurna. Los pacientes se quejan de presentar una o varias siestas involuntarias.

Esto es consecuencia de que se privan de sueño de manera crónica e intencional.

Este trastorno es muy frecuente en adolescentes y adultos jóvenes.

---

###Ronquido

El ronquido es uno de los trastornos más frecuentes del sueño, en especial en personas mayores de 40 años; puede ser la alerta para un problema de salud mayor, es un problema complejo que incluye varios factores de riesgo.

El principal factor de riesgo en la población para presentar ronquido es tener sobre peso u obesidad.

Las complicaciones de la enfermedad son de varios tipos:

* rendimiento no óptimo del individuo.
* riesgo de cometer errores durante nuestras actividades cotidianas.
* probabilidad de que haya accidentes laborales o vehiculares.

Incluso, se corre el riesgo de desarrollar hipertensión o arritmias.

---

###Narcolepsia

Es una enfermedad rara de causa desconocida que usualmente aparece en la adolescencia y juventud.

Se caracteriza por periodos de sueño incontrolable, perdida de la fuerza inmediatamente después de una emoción intensa, otro síntoma característico es la cataplejía, (perdida brusca del tono muscular). Su duración suele ser de segundos o pocos minutos y la recuperación es inmediata y completa.

Otros síntomas:

* Alucinaciones hipnagógicas. (sueños muy vivenciados que ocurren especialmente al inicio del sueño, y que cursan con fenómenos visuales, táctiles y auditivos)

* Sueño desorganizado o conductas automáticas.

---

###Parasomnias.

Son conductas indeseables que ocurren durante el sueño como hablar, llorar, comer, gritar o caminar. Aunque la parasomnia más común son las pesadillas (sueños emocionalmente intensos y desagradables).

Lo habitual es que aparezcan durante la infancia y disminuyan en la adolescencia.

Para prevenirlas es necesario tener buenos hábitos antes de dormir. Entre las parasomnias más conocidas están:

* Sonambulismo.
* Orinarse dormido.

* Terrores nocturnos. (Se refiere a esa sensación de terror durante la noche, acompañada de patadas, sonidos con la boca que otros no pueden entender y respiración convulsa.)

* Conducta violenta durante el sueño.
* Conducta sexual durante el sueño.
* Parálisis de sueño.

---

###Sonambulismo.

Es un tipo de **parasomnia**. Son movimientos físicos o conductas no deseables o molestas, que pasan sobre todo o solo durante el sueño.

El niño o adulto se levanta de la cama y camina sin ser completamente consciente de lo que le rodea. Llegan a hacer varias cosas, desde sentarse dormidos hasta caminar y correr muy agitados, así como realizar tareas complejas como abrir puertas cerradas con llave, sacar alimentos del refrigerador, bajar las escaleras, e incluso salir de casa.

Suele mantener los ojos abiertos y pueden murmurar o dar respuestas sin sentido. Es difícil despertarle y cuando termina el episodio se vuelven a dormir plácidamente. Es raro que el individuo recuerde lo que pasó al día siguiente.

##Modelo

El modelo decidido, dependerá de la caracteristica de los datos, por lo ya visto en el EDA, tenemos datos categóricos y numéricos a su vez, los datos numéricos tienen escalas diferentes, como las diferentes categorias están desbalanceadas, las cuales son 3 en este caso se trata de una clasificación multivariable, por lo que con todo lo mencionado, el modelo más adecuado, seria un random forest, es robusto ante la escala, puede manejar datos categoricos y numéricos (así evitandonos realizar un onehotencoder que crearía más variables y aumentaria el costo computacional).

Por lo que vamos a proceder a ajustar este modelo

```python
df.head()
```

```python
#vamos a quitar "PersonID" dado que es irrelevante.
df.drop('Person ID', axis=1, inplace=True)
```

```python
#y quitaremos la columna "BMI Numeric" que creamos para una gráfica
df.drop('BMI Numeric', axis=1, inplace=True)
```

```python
df.head()
```

```python
#necesitamos dividir la columna "blood pressure" dado que la diagonal hace que los datos de esta columna sean tomados como tipo obj
#y meter un onehotencoder a los datos originales podria sobreajustar el modelo ya que el modelo aprendería a clasificar datos como 126/83 o 125/80
#pero si le llega un valor nuevo como 124/81 pues no sabría clasificarlo.
#Las columnas resultantes, serán tipo numérica por que no hace falta realizar más modificaciones a esta columna.

df = pd.concat([df, df['Blood Pressure'].str.split('/', expand=True)], axis=1).drop('Blood Pressure', axis=1)
df = df.rename(columns={0: 'Systolic', 1: 'Diastolic'})
df['Systolic'] = df['Systolic'].astype(float)
df['Diastolic'] = df['Diastolic'].astype(float)

#Seleccionamos Columnas categóricas
categorical_cols = ['Gender', 'Occupation', 'BMI Category']

#Tomamos nuestra variable de interés
le = LabelEncoder()
y = le.fit_transform(df['Sleep Disorder'])

#y la excluimos, pues es la que nos interesa
exclude_cols = ['Sleep Disorder']

#Definimos transformador
preprocessor = ColumnTransformer(
    transformers=[
        ('num', 'passthrough', df.drop(categorical_cols + exclude_cols, axis=1).columns),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
    ])

X = preprocessor.fit_transform(df.drop(exclude_cols, axis=1))

#Dividimos los datos, un 70/30
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

#Entrenamos el Random Forest
rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    min_samples_split=5,
    random_state=42
)
rf.fit(X_train, y_train)

#Evaluamos
y_pred = rf.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred, target_names=le.classes_))

#Creamos una Matriz de confusión
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=le.classes_,
            yticklabels=le.classes_)
plt.title('Matriz de Confusión')
plt.xlabel('Predicho')
plt.ylabel('Real')
plt.show()
```

En el caso de **Insomnio**, tuvimos 20 pacientes correctamente clasificados y 4 Falsos negativos (2 en no transtorno y 2 en Apnea del sueño)
El modelo toma algunos casos de insomnio como Apneas o como ausencia de transtorno del sueño.

En el caso de **No transtorno**, 61 pacientes fueron correctamente identificados, y 1 falso negativo, ("tomado como insomnio)
Considero es un buen modelo para identificar posibles casos sin algun transtorno del sueño.

En el caso de **Apnea del sueño** Clasificamos correctamente a 20 pacientes y tuvimos 7 falsos negativos (5 como insomnio y 2 como "No transtorno")

Específicamente en Apnea es donde más falla, pues confunde ciertos casos de Apnea del sueño con insomnio.

##Conclusión

En conclusion, el modelo muestra buenos resultados, puesto que aunque tiene problemas de clasificación la cantidad de errores Tipo 1 y Tipo 2, es mínima, además de que las métricas de error de cada una de las clasificaciones son bastante altas, esto quiere decir que el modelo tiene buen desempeño.

Tratar el desbalanceo de datos, puesto que a pesar de que Random forest es un modelo bastante robusto ante el desbalanceo de datos, lo mejor sería presentar el menor desbalanceo posible en los datos, para que el modelo tenga un mejor desempeño.

Además de buscar mejores hiperparámetros para el modelo.
